```python
# pandas series and DataFrame Operations

!pip install pandas
import pandas as pd

#....................SERIES....................

# creating a series
series1 = pd.Series([10,20,30,40,50])

print("Pandas series")
print(series1)

# Accessing series Elements
print("\nfirst Element:")
print(series1[0])

# Series with custom Index
series2 = pd.Series(
    [85,90,95],
    index = ["Math", "Science", "English"]
)

print("\nseries with custom index:")
print(series2)

#....................DATAFRAME....................

# creating DataFrame using Dictionary
student = {
    "ID": [101,102,103],
    "Name": ["sirushti", "varsha", "sadhana"],
    "Marks": [95,88,90]
}

df = pd.DataFrame(student)

print("\nDataFrame:")
print(df)

# Display first rows
print("\nFirst rows:")
print(df.head())

# Display last row
print("\nLast rows:")
print(df.tail())

# Display column names
print("\nColumn names:")
print(df.columns)

# Access single column
print("\nName column:")
print(df["Name"])

# Access multiple columns
print("\nName and Marks columns:")
print(df[["Name", "Marks"]])

# Access rows using loc
print("\nRow using loc:")
print(df.loc[0])

# Access rows using iloc
print("\nRow using iloc:")
print(df.iloc[1])

# Add a new column
df["Grade"] = ["A", "B", "A"]

print("\nDataFrame after adding Grade:")
print(df)

# Update a specific cell
df.loc[1, "Marks"] = 92

print("\nDataFrame after updating Marks:")
print(df)

# Remove a column
df = df.drop("Grade", axis=1)

print("\nDataFrame after removing Grade:")
print(df)

# DataFrame information
print("\nDataFrame Information:")
print(df.info())

# Statistical Summary
print("\nStatistical Summary:")
print(df.describe())
```

# PRACTICAL 12

```python
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame({
    "Name": ["Sirushti", "Thansi", "Sumathi", "Elim"],
    "Marks": [90, np.nan, 80, np.nan],
    "Age": [18, 19, np.nan, 20]
})

print("Original DataFrame:\n", df)

# Check missing values
print("\nMissing values:\n", df.isnull())

# Count missing values
print("\nMissing count:\n", df.isnull().sum())

# Fill missing values with 0
print("\nFill with 0:\n", df.fillna(0))

# Fill Marks with mean
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
print("\nMarks filled with mean:\n", df)

# Forward fill
print("\nForward fill:\n", df.ffill())

# Backward fill
print("\nBackward fill:\n", df.bfill())

# Remove rows with missing values
print("\nAfter dropna:\n", df.dropna())

# Replace a value
df = df.replace(80, 85)
print("\nAfter replace:\n", df)

# Display information
print("\nDataFrame Information:")
df.info()
```

```python
# Grouping and aggregating data using pandas

import pandas as pd

# Creating DataFrame 
data = {
    "Department": ["HR","IT","HR","IT","Sales","Sales"],
    "Employee": ["sumathi", "Thansi","Jeslyn","Sirushti","Sam","Prince"],
    "Salary":[30000,50000,35000,60000,45000,40000],
    "Experience":[2,5,3,6,4,2]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Grouping by Department
grouped = df.groupby("Department")

# Sum Aggregation
print("\nSum of Salary by Department:")
print(grouped["Salary"].sum())

# Mean Aggregation
print("\nAverage Salary by Department:")
print(grouped["Salary"].mean())

# Maximum Salary
print("\nMaximum Salary by Department:")
print(grouped["Salary"].max())

# Minimum Salary
print("\nMinimum Salary by Department:")
print(grouped["Salary"].min())

# Multiple Aggregations
print("\nMultiple Aggregations:")
print(
    grouped["Salary"].agg(
        ["sum", "mean", "max", "min"]
    )
)

# Grouping Multiple Columns
print("\nGrouping by Department and Experience:")

group_multi = df.groupby(
    ["Department", "Experience"]
)["Salary"].sum()

print(group_multi)

# Iterating Groups
print("\nDisplaying Each Group:")

for name, group in grouped:
    print("\nDepartment:", name)
    print(group)

# Sorting Values
print("\nSorted DataFrame by Salary:")

sorted_df = df.sort_values(by="Salary")
print(sorted_df)
```

```python
# Merge and Join DataFrames using Pandas

import pandas as pd

# ----- CONCATENATE OPERATIONS ------------------- (all columns name should be same)

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])   

print("\nDataFrame 1:")
print(df1)

#  Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [201, 202, 203, 205],
    "Marks": [85, 78, 80, 65]
}, index=[4, 5, 6, 7])   

print("\nDataFrame 2:")
print(df2)

frames = [df1,df2]
result = pd.concat(frames)
print("concatenate dataframes")
print(result)

# ---------------- MERGE OPERATIONS ---------------- ( at least one column should be same, others can be different  )

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Name": ["Sirushti", "varsha", "Akshaya", "sadhana"]
})

print("DataFrame 1:")
print(df1)

# Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])   

print("\nDataFrame 2:")
print(df2)

# Inner Merge
inner_merge = pd.merge(df1, df2, on="ID", how="inner")

print("\nInner Merge:")
print(inner_merge)

# Left Merge
left_merge = pd.merge(df1, df2, on="ID", how="left")

print("\nLeft Merge:")
print(left_merge)

# Right Merge
right_merge = pd.merge(df1, df2, on="ID", how="right")

print("\nRight Merge:")
print(right_merge)

# Outer Merge
outer_merge = pd.merge(df1, df2, on="ID", how="outer")

print("\nOuter Merge:")
print(outer_merge)

# ---------------- JOIN OPERATIONS ---------------- ( columns names in both datframes is different )

# Creating DataFrames for Join
df3 = pd.DataFrame({"Thansi","Sumathi","Elim"}, 
index=[1, 2, 3])

df4 = pd.DataFrame({
    "Salary": [30000, 40000, 50000]
}, index=[1, 2, 3])

print("\nDataFrame 3:")
print(df3)

print("\nDataFrame 4:")
print(df4)

# Joining DataFrames
joined_df = df3.join(df4)

print("\nJoined DataFrame:")
print(joined_df)

# Joining with Different Join Types
left_join = df3.join(df4, how="left")

print("\nLeft Join:")
print(left_join)

right_join = df3.join(df4, how="right")

print("\nRight Join:")
print(right_join)

outer_join = df3.join(df4, how="outer")

print("\nOuter Join:")
print(outer_join)
```

```python
# Simple Plots using Matplotlib

import matplotlib.pyplot as plt 

#..................LINE PLOT.....................

x = [1,2,3,4,5]
y = [10,20,30,40,50]

plt.figure(figsize=(5,4))

plt.plot(x,y)

plt.title("Line plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()

#....................BAR GRAPH.....................

students = ["A", "B","C","D"]
marks = [85,90,75,88]

plt.figure(figsize=(5,4))

plt.bar(students, marks)

plt.title("Bar Graph")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()

#.....................HISTOGRAM.....................

data = [10,20,20,30,30,30,40,50,50]

plt.figure(figsize=(5,4))

plt.hist(data,bins=5)

plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

```python
# Data Visualization using Seaborn

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Creating Sample Dataset
data = {
    "Maths": [85, 90, 78, 92, 88],
    "Science": [80, 85, 75, 95, 89],
    "English": [78, 83, 72, 90, 84]
}

# Creating DataFrame
df = pd.DataFrame(data)

print("Dataset:")
print(df)

# -------------------- HEATMAP --------------------

plt.figure(figsize=(5,4))

sns.heatmap(df, annot=True)

plt.title("HeatMap")

plt.show()

#......................BOXPLOT.......................

plt.figure(figsize=(5,4))

sns.boxplot(data=df)

plt.title("Boxplot")

plt.show()

#...................PAIRPLOT...........................

sns.pairplot(df)

plt.show()
```
