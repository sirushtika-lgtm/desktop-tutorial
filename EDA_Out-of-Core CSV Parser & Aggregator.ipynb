```python
chunksize=10000
```

```python
total_rows = 0
total_sum = None
```

```python
for chunk in pd.read_csv(r"C:\Users\Sirushtika S\OneDrive\Desktop\weekly task\transaction.csv.csv", chunksize=chunksize):
    total_rows += len(chunk)
    chunk_sum = chunk.select_dtypes(include='number').sum()
    if total_sum is None:
        total_sum = chunk_sum
    else:
        total_sum = total_sum.add(chunk_sum, fill_value=0)
```
